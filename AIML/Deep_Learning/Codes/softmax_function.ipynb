```python
import numpy as np
```

```python
def softmax(z):
    return np.exp(z) / np.sum(np.exp(z), axis=0)
```

```python
z = np.array([[-1, 3, 10],
             [5, 2, -4],
             [3,-5,15]])
s = softmax(z)
s
```

```python
def softmax1(Z):
    Z_exp = np.exp(Z - np.max(Z, axis=0))
    return Z_exp / np.sum(Z_exp, axis=0)
```

```python
p = softmax1(z)
print(p)    
```

```python
y = np.array([2, 1, 0])
y
```

```python
print(p[y, np.arange(p.shape[1])])
```
