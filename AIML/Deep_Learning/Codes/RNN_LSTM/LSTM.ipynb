```python
# Construct time series from a multiple sequences
def prep_timeseries2(x, y, n_steps=1):
    x_ts, y_ts = [], []
    for i in range(len(x) - n_steps):
        x_ts.append(x.iloc[i: (i+n_steps)].values)
        y_ts.append(y.iloc[i+n_steps])
    return np.array(x_ts), np.array(y_ts)
```

```python
## User defined function to prepare the univariate time series data
def prep_timeseries(x, n_steps = 1):
    x_ts, y_ts = [], []
    for i in range(len(x) - n_steps):
        x_ts.append(x.iloc[i: (i+n_steps)].values)
        y_ts.append(x.iloc[i+n_steps])
    return np.array(x_ts), np.array(y_ts)

```
