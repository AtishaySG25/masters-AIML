```python
import pandas as pd
import numpy as np
import os
import sys
import sympy as sp
import matplotlib.pyplot as plt
import matplotlib.cm as cm
plt.style.use('seaborn-whitegrid')
%matplotlib inline
```

```python
# Markov Process
# Probability transition marix (one-step transition matrix)
T = np.array([[0.95, 0.05, 0, 0], [0.04, 0.85, 0.1, 0.01], [0, 0, 1, 0], [0, 0, 0, 1]])

# System dynamics
# Time period
t_end = 90

# Starting state of the system is susceptible
s = np.empty(t_end).astype('int32')
s[0] = 0 # susceptible state at the beginning

# Evolution of the system
for j in range(1, t_end):
  s[j] = np.random.choice([0, 1, 2, 3], p = T[s[j-1]])
  #print(s[j])

fig, ax = plt.subplots(1, 1, figsize = (6,6))
fig.tight_layout(pad = 4.0)
ax.plot(s, c = 'blue', linewidth = 2.5)
ax.set_xlabel('Time in days')
ax.yaxis.set_ticks([0, 1, 2, 3])
ax.set_yticklabels(['Susceptible', 'Infected', 'Recovered', 'Deceased'], rotation = 0);
```
